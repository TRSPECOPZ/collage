```r
# Import the CSV as a data frame

students <- read.csv("Project_Student_Results.csv")

head(students)

# Show the first six rows

dim(students)

# Show the number of rows and columns

names(students)

# List column names

str(students)

# Show variable types and sample values

summary(students)

# Summarise each variable
```

```r
unique(students$Programme)

# List distinct programme labels

unique(students$StudyMode)

# List distinct study mode labels

table(students$Programme)

# Count students in each programme

table(students$StudyMode)

# Count students in each study modes
```

```r
clean <- students
# trim leading or trailing spaces from categorical labels, then inspectthem again:

clean$Programme <- trimws(clean$Programme)

# Remove spaces at thestart/end of labels (trimws – trim whitespace

clean$StudyMode <- trimws(clean$StudyMode)

# Remove spaces at thestart/end of labels

unique(clean$Programme)

# Check programme labels after trimming

unique(clean$StudyMode)

# Check study mode labels after trimming
```

```r
# Replace blank StudyMode entries with NA

clean$StudyMode[trimws(clean$StudyMode) == ""] <- NA

# Change lowercase "online" to "Online"

clean$StudyMode[clean$StudyMode == "online"] <- "Online"

# Change "On-campus" to "On campus"

clean$StudyMode[clean$StudyMode == "On-campus"] <- "On

campus"

unique(clean$StudyMode)

# Replace negative commuting times with NA

clean$CommuteMinutes[!is.na(clean$CommuteMinutes) &

clean$CommuteMinutes < 0] <- NA

summary(clean$CommuteMinutes)
```

```r
# Check that the cleaned dataset is available

dim(clean)

# Show rows and columns

names(clean)

# Check column names

summary(clean)

# Review the cleaned variables
```

```r
# Summary statistics for weekly study hours

mean(clean$StudyHours, na.rm = TRUE)

# Mean

median(clean$StudyHours, na.rm = TRUE)

# Median

sd(clean$StudyHours, na.rm = TRUE)

# Standard deviation

range(clean$StudyHours, na.rm = TRUE)

# Minimum and maximum

IQR(clean$StudyHours, na.rm = TRUE)

# Interquartile range
```

```r
# Summary statistics for sleep hours

mean(clean$SleepHours, na.rm = TRUE)

# Mean
median(clean$SleepHours, na.rm = TRUE)

# Median

sd(clean$SleepHours, na.rm = TRUE)

# Standard deviation

range(clean$SleepHours, na.rm = TRUE)

# Minimum and maximum

IQR(clean$SleepHours, na.rm = TRUE)

# Interquartile range
```

```r
# Histogram: frequency distribution of sleep hours

hist(clean$SleepHours, main = "Distribution of Sleep Hours",

xlab = "Sleep hours", col = "lightblue")

# Boxplot: median, IQR, spread and potential outliers

boxplot(clean$SleepHours, main = "Sleep Hours",

ylab = "Sleep hours", col = "lightblue")
```

```r
# Count students in each programme;show missing values if present

table(clean$Programme, useNA = "ifany")

# Bar chart of recorded programmes

barplot(table(clean$Programme), main = "Students by Programme",

xlab = "Programme", ylab = "Number of students",

col = "steelblue")
```

```r
# Plot attendance against assignment marks

plot(clean$Attendance, clean$AssignmentMark,

main = "Attendance and Assignment Marks",

xlab = "Attendance (%)",

ylab = "Assignment Mark (%)",

pch = 19, col = "steelblue")

# Calculate Pearson's correlation

cor(clean$Attendance, clean$AssignmentMark,

use = "complete.obs", method = "pearson")
```

```r
# Compare assignment marks across study modes

boxplot(AssignmentMark ~ StudyMode, data = clean,

main = "Assignment Marks by Study Mode",

xlab = "Study Mode", ylab = "Assignment Mark (%)",

col = "lightblue")

# Grouped summary statistics

aggregate(AssignmentMark ~ StudyMode, data = clean, FUN = mean)

aggregate(AssignmentMark ~ StudyMode, data = clean, FUN = median)

aggregate(AssignmentMark ~ StudyMode, data = clean, FUN = sd)

aggregate(AssignmentMark ~ StudyMode, data = clean, FUN = IQR)
```
